```python
import geopandas as gpd
import pandas as pd
from pysits import *
import os
```

```python

# lê o gpkg, reprojeta pra WGS84, renomeia e adiciona colunas de data
amostras_sf = (
    gpd.read_file("/home/jovyan/SR_agricola/feijao/dados/pontos_feijao.gpkg")
    .to_crs(epsg=4326)
    .rename(columns={"Class_s": "label"})
)

amostras_sf["start_date"] = pd.to_datetime("2023-09-30")  # ajuste ao início real da safra de feijão
amostras_sf["end_date"] = pd.to_datetime("2024-09-29")    # ajuste ao fim real do ciclo

# mantém só o essencial + geometria
amostras_sf = amostras_sf[["label", "start_date", "end_date", "geometry"]]

# filtra só feijão
amostras_feijao = amostras_sf[amostras_sf["label"] == "Feijao"]

# extrai coordenadas
lon_max = amostras_feijao.geometry.x.max()
lon_min = amostras_feijao.geometry.x.min()
lat_max = amostras_feijao.geometry.y.max()
lat_min = amostras_feijao.geometry.y.min()

# define o roi para os dados de feijão
roi_feijao = {
    "lat_max": lat_max,
    "lat_min": lat_min,
    "lon_max": lon_max,
    "lon_min": lon_min,
}
```

```python
print(amostras_feijao)
amostras_feijao.plot()
```

```python
roi_feijao
```

```python
# Define a data cube in the BDC repository based on the LEM ROI
bdc_cube = sits_cube(
    source = "BDC", 
    collection  = "SENTINEL-2-16D",
    bands = ("NDVI", "EVI"),
    roi = roi_feijao,
    start_date = "2023-09-30",
    end_date = "2024-09-29"
)
```

```python
tempdir_py = "/home/jovyan/SR_agricola/feijao/tempdir"
os.makedirs(tempdir_py, exist_ok=True)

# Copy the region of interest to a local directory
lem_cube = sits_cube_copy(
    cube = bdc_cube,
    roi = roi_feijao,
    output_dir = tempdir_py
)
```

```python
plot(lem_cube, tile = "023032", band = "NDVI", date = "2023-09-30")
plot(lem_cube, tile = "023033", band = "NDVI", date = "2023-09-30")
```

```python
lem_cube
```

```python
print(sits_timeline(lem_cube))

print(len(sits_timeline(lem_cube)))
```

```python
amostras_feijao.loc[:, "longitude"] = amostras_feijao.geometry.x
amostras_feijao.loc[:, "latitude"] = amostras_feijao.geometry.y
amostras_feijao = amostras_feijao.copy()
amostras_feijao.rename(columns={"label": "class"}, inplace=True)
amostras_feijao
```

```python
samples_lem_time_series = sits_get_data(
    cube = lem_cube,
    samples = amostras_feijao
)
```

```python
summary(samples_lem_time_series)
```

```python
# Select all samples with label "Cropland_2_cycles" using `query` 
# since the label attribute is a column of the samples data.frame
samples_crops_2cycles = samples_lem_time_series.query('label == "NoClass"')

# Select the NDVI band values using sits_select
# because band values are in a nested data.frame
samples_crops_2cycles_ndvi = sits_select(samples_crops_2cycles, bands = "NDVI")

# plot the samples for label Cropland_2_cycles and band NDVI
plot(samples_crops_2cycles_ndvi)
```

# Usando STAC

```python
import pystac_client
import os
```

```python
service = pystac_client.Client.open('https://data.inpe.br/bdc/stac/v1/')
```

```python
collection = service.get_collection('S2-16D-2')
collection
```

```python
b = (-25.63763272049445, -26.086748824232362, -53.05931164056312, -53.34543525919164)
bb = (-53.35543525919164, -26.096748824232362, -53.04931164056312, -25.62763272049445)

item_search = service.search(bbox=bb,
                             datetime='2023-09-30/2024-09-29',
                             collections=['S2-16D-2'])
item_search
```
